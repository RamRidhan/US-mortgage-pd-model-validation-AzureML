```python
import sys; sys.path.insert(0, '../src')
import os; os.chdir('..')
import pandas as pd
from load import load_orig, load_perf
from snapshots import add_default_event, build_snapshot
```

```python
orig = load_orig()
perf = add_default_event(load_perf())
print(orig.shape, perf.shape)
```

```python
out = []
for y in range(2015, 2020):
    s, n_all = build_snapshot(perf, f'{y}-12-01', orig)
    print(y, len(s), s.default_12m.sum(), f'{s.default_12m.mean():.4%}')
    out.append(s)
pd.concat(out, ignore_index=True).to_parquet('data/interim/snapshots.parquet')
```
